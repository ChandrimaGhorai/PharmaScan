```python
# ==========================================
# PharmaScan: Drug-Likeness Screening Engine
# (User-Friendly Chemical Structure Input)
# ==========================================

from rdkit import Chem
from rdkit.Chem import Descriptors, Crippen

def evaluate_drug_likeness(smiles, compound_name="Custom Compound"):
    mol = Chem.MolFromSmiles(smiles)
    if not mol:
        print(f"\n[Error]: Could not parse the provided chemical structure for '{compound_name}'. Please ensure valid notation.")
        return

    # Lipinski Descriptors
    mw = Descriptors.MolWt(mol)
    logp = Crippen.MolLogP(mol)
    hbd = Descriptors.NumHDonors(mol)
    hba = Descriptors.NumHAcceptors(mol)

    # Veber Descriptors (Flexibility & Polarity)
    tpsa = Descriptors.TPSA(mol)
    rot_bonds = Descriptors.NumRotatableBonds(mol)

    # Rule Violation Tracking
    violations = 0
    violation_details = []

    if mw > 500:
        violations += 1
        violation_details.append(f"Molecular Weight ({mw:.2f} Da) exceeds 500")
    if logp > 5:
        violations += 1
        violation_details.append(f"LogP ({logp:.2f}) exceeds 5")
    if hbd > 5:
        violations += 1
        violation_details.append(f"Hydrogen Bond Donors ({hbd}) exceed 5")
    if hba > 10:
        violations += 1
        violation_details.append(f"Hydrogen Bond Acceptors ({hba}) exceed 10")
    if tpsa > 140:
        violations += 1
        violation_details.append(f"Topological Polar Surface Area ({tpsa:.2f} Å²) exceeds 140 Å²")
    if rot_bonds > 10:
        violations += 1
        violation_details.append(f"Rotatable Bonds ({rot_bonds}) exceed 10")

    # Classification verdict
    status = "Favorable Drug-Likeness (Likely Orally Bioavailable)" if violations <= 1 else "Poor Bioavailability Risk (Multiple Rule Violations)"

    print("\n" + "="*55)
    print(f"PHARMASCAN SCREENING REPORT: {compound_name}")
    print("="*55)
    print(f"Chemical Structure (SMILES)   : {smiles}")
    print(f"Molecular Weight (MW)         : {mw:.2f} Da (Limit: <= 500)")
    print(f"Lipophilicity (LogP)          : {logp:.2f}   (Limit: <= 5)")
    print(f"Hydrogen Bond Donors (HBD)    : {hbd}       (Limit: <= 5)")
    print(f"Hydrogen Bond Acceptors (HBA) : {hba}      (Limit: <= 10)")
    print(f"Polar Surface Area (TPSA)     : {tpsa:.2f} Å²(Limit: <= 140)")
    print(f"Rotatable Bonds               : {rot_bonds}     (Limit: <= 10)")
    print("-" * 55)
    print(f"Total Rule Violations         : {violations}")
    if violation_details:
        for detail in violation_details:
            print(f"  - Warning: {detail}")
    print(f"Screening Verdict             : {status}")
    print("="*55)

def main():
    print("=== PharmaScan: Cheminformatics Control Console ===")
    print("Select Analysis Option:")
    print("1. Screen Preset Benchmark Drug (e.g., Aspirin, Ibuprofen, Cyclosporin A)")
    print("2. Enter Custom Chemical Structure")

    choice = input("\nEnter choice (1 or 2): ").strip()

    preset_db = {
        "1": {"name": "Aspirin", "smiles": "CC(=O)Oc1ccccc1C(=O)O"},
        "2": {"name": "Ibuprofen", "smiles": "CC(C)Cc1ccc(cc1)C(C)C(=O)O"},
        "3": {"name": "Cyclosporin A (Large Molecule)", "smiles": "CCC1C(=O)N(CC(=O)N(C)CC(=O)N(C)CC(=O)N(C)CC(=O)N(C)C(C(=O)N(C)CC(=O)N(C)C(C(=O)N(C)C(C(=O)N(C)C(C(=O)N1)C(C)CC=CC)C)C(C)CC)CC(C)C)C"}
    }

    if choice == "1":
        print("\nSelect preset compound:")
        for k, v in preset_db.items():
            print(f"[{k}] {v['name']}")
        p_choice = input("Select option (1, 2, or 3): ").strip()
        selected = preset_db.get(p_choice, preset_db["1"])
        evaluate_drug_likeness(selected["smiles"], selected["name"])

    elif choice == "2":
        print("\nTip: You can use standard notation format (e.g., Paracetamol: CC(=O)NC1=CC=C(C=C1)O)")
        custom_structure = input("Enter Chemical Structure: ").strip()
        custom_name = input("Enter compound identifier/name: ").strip() or "Custom Compound"
        evaluate_drug_likeness(custom_structure, custom_name)
    else:
        print("[Error]: Invalid selection.")

if __name__ == "__main__":
    main()
```
